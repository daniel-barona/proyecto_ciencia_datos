#ELIMINAR TILDES Y PASAR TODOS LOS REGISTROS A MINUSCULAS

```python
# Carpeta de salida
trusted_dir = "/content/proyecto_ciencia_datos/data/trusted"
os.makedirs(trusted_dir, exist_ok=True)
# Función para normalizar texto
def normalizar_texto(valor):

    if pd.isna(valor):
        return valor

    if not isinstance(valor, str):
        return valor

    # Eliminar tildes
    valor = "".join(
        c
        for c in unicodedata.normalize("NFD", valor)
        if unicodedata.category(c) != "Mn"
    )

    # Convertir a minúsculas
    valor = valor.lower().strip()

    return valor
# Aplicar solo a columnas de texto
columnas_texto = df.select_dtypes(include="object").columns

for columna in columnas_texto:
    df[columna] = df[columna].apply(normalizar_texto)
# Guardar archivo
ruta_salida = os.path.join(
    trusted_dir,
    "SIPSA_2013_2026_trusted.xlsx"
)

df.to_excel(
    ruta_salida,
    index=False
)

print("Proceso finalizado correctamente.")
print(f"Archivo guardado en:\n{ruta_salida}")
```

#ELIMINAR PUNTOS AL FINAL DEL MERCADO

```python
# Carpeta de salida
trusted_dir = "/content/proyecto_ciencia_datos/data/trusted"
os.makedirs(trusted_dir, exist_ok=True)
# Eliminar punto final de los mercados
mercados_antes = df["Mercado"].copy()

df["Mercado"] = (
    df["Mercado"]
      .astype(str)
      .str.strip()
      .str.rstrip(".")
      .str.strip()
)
# Mostrar resultados
registros_modificados = (mercados_antes != df["Mercado"]).sum()

print("=" * 80)
print(f"Registros modificados: {registros_modificados:,}")
print("=" * 80)
# Guardar archivo
ruta_salida = os.path.join(
    trusted_dir,
    "SIPSA_2013_2026_trusted.xlsx"
)
df.to_excel(
    ruta_salida,
    index=False
)
print("\nProceso finalizado correctamente.")
print(f"Archivo guardado en:\n{ruta_salida}")
```

#AJUSTE DE LAS COLUMNAS (CAMBIAR ESPACIOS POR '_' )

Este ajuste se realiza con el fin de mejorar las buenas practicas para de las columnas (variables) para tener un buen manejo en la hora de realizar el codigo

```python
# Renombrar columnas
df = df.rename(columns={
    "Fecha": "fecha",
    "Grupo": "grupo",
    "Producto": "producto",
    "Departamento": "departamento",
    "Código departamento": "codigo_departamento",
    "Municipio": "municipio",
    "Código municipio": "codigo_municipio",
    "Mercado": "mercado",
    "Precio promedio por kilogramo*": "precio_promedio_kg",
    "Circulación" : "circulacion"
})
# Mostrar nombres de columnas
print("Columnas actualizadas:\n")
print(df.columns.tolist())
# Carpeta de salida
trusted_dir = "/content/proyecto_ciencia_datos/data/trusted"
os.makedirs(trusted_dir, exist_ok=True)
# Guardar archivo
ruta_salida = os.path.join(
    trusted_dir,
    "SIPSA_2013_2026_trusted.xlsx"
)
df.to_excel(
    ruta_salida,
    index=False
)
print("\n" + "=" * 80)
print("Proceso finalizado correctamente.")
print(f"Archivo guardado en:\n{ruta_salida}")
print("=" * 80)
```

#ELIMINAR LOS ESPACIOS DOBLES

```python
trusted_dir = "/content/proyecto_ciencia_datos/data/trusted"
os.makedirs(trusted_dir, exist_ok=True)

# Eliminar espacios dobles en todas las columnas de texto

columnas_texto = df.select_dtypes(include="object").columns

total_modificados = 0

for columna in columnas_texto:

    antes = df[columna].copy()

    df[columna] = (
        df[columna]
          .astype(str)
          .str.replace(r"\s+", " ", regex=True)
          .str.strip()
    )

    total_modificados += (antes != df[columna]).sum()

print("=" * 80)
print(f"Registros modificados: {total_modificados:,}")
print("=" * 80)

# Guardar archivo actualizado
ruta_salida = os.path.join(
    trusted_dir,
    "SIPSA_2013_2026_trusted.xlsx"
)

df.to_excel(
    ruta_salida,
    index=False
)

print("\nProceso finalizado correctamente.")
print(f"Archivo guardado en:\n{ruta_salida}")
```
